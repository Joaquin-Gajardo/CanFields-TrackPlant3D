```python
import pandas as pd
```

```python
# Load data
df = pd.read_csv('../results_test_organs_epoch15000.csv', index_col=0)
df
```

### Select method

```python
method = "TPSR"
method_subtable = df.loc[df['method'] == method].reset_index(drop=True)
```

#### Checking for duplicated (SOLVED)

```python

method_subtable.sort_values(by=['sequence', 'experiment_timestamp'], inplace=True, ignore_index=True)
is_duplicated = method_subtable.duplicated(subset=['sequence'], keep=False)
method_subtable
```

```python
method_subtable.loc[is_duplicated, :].reset_index(drop=True)
```

```python
method_subtable.loc[~is_duplicated, :].reset_index(drop=True)
```

```python
method_subtable['timestamp'] = method_subtable['experiment'].apply(lambda x: '_'.join(x.split('_')[:2]))
method_subtable.sort_values(by=['sequence', 'timestamp'], inplace=True, ignore_index=True)
method_subtable
```

#### Check epoch used

```python
method_subtable['epoch'].value_counts()
```

```python
method_subtable.loc[method_subtable['epoch'] != 15000, :]
```

### Find missing runs in all methods

```python
# Analysis: Missing sequences per method
all_sequences = set(df['sequence'].unique())
all_methods = sorted(df['method'].unique())

for method in all_methods:
    method_sequences = set(df[df['method'] == method]['sequence'].unique())
    missing = sorted(all_sequences - method_sequences)
    print(f"{method}: {len(method_sequences)}/{len(all_sequences)} sequences")
    if missing:
        print(f"  Missing: {missing}")
```
